2️⃣ Filter rows, group, aggregate, then filter groups (one chained expression)

Example:

Keep only rows where FILTER_COLUMN == SOME_VALUE

Group by GROUP_COLUMN

Take the mean of NUM_COLUMN

Keep only groups where that mean is ≥ some threshold

```python
subset_grouped_filtered = (
    DF[DF["FILTER_COLUMN"] == FILTER_VALUE]     # keep only rows matching your condition
      .groupby("GROUP_COLUMN", as_index=False)  # group by this column
      .agg(MEAN_NUM=("NUM_COLUMN", "mean"))     # compute mean of NUM_COLUMN for each group
      .loc[lambda d: d["MEAN_NUM"] >= THRESHOLD_VALUE]  # keep groups with mean >= threshold
)
# - Replace FILTER_VALUE with a value (e.g., "Yes" or 10)
# - Replace THRESHOLD_VALUE with a number (e.g., 50)
# This entire block is ONE chained expression (counts as "one line" logically).
```

3️⃣ Recode a numeric variable into categories (string labels)

Example:

Suppose SCORE is numeric

Create SCORE_LEVEL as "low", "medium", "high" based on intervals

```python
DF["SCORE_LEVEL"] = pd.cut(
    DF["NUMERIC_COLUMN"],                    # numeric column you are recoding
    bins=[MIN_VALUE, MID1, MID2, MAX_VALUE], # boundaries of your intervals
    labels=["LOW", "MEDIUM", "HIGH"],        # labels for each interval
    include_lowest=True                      # include the very smallest value
)
# Example: bins=[0, 50, 80, 100] and labels=["LOW", "MEDIUM", "HIGH"]
```

4️⃣ Group, aggregate, then sort by an aggregated stat (descending, chained)

Example:

Group by GROUP_COLUMN

Compute mean of NUM_COLUMN and sum of OTHER_NUM_COLUMN

Sort by the mean in descending order

```python
grouped_sorted = (
    DF
      .groupby("GROUP_COLUMN", as_index=False)              # group by this column
      .agg(
          MEAN_NUM=("NUM_COLUMN", "mean"),                  # mean of NUM_COLUMN
          SUM_OTHER=("OTHER_NUM_COLUMN", "sum")             # sum of OTHER_NUM_COLUMN (optional)
      )
      .sort_values("MEAN_NUM", ascending=False)             # sort by MEAN_NUM, largest first
)
# Again, this is one chained expression.
```
