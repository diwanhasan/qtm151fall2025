How would you change the code below, so that it prints out the position in the list of each non-numerical item it encounters?

Implement those changes, and run the code cell

```python
### Make changes to the code below

list_mixed = [1, 2, None, None, 4, 11, "Apple", 2.2, 13, "!!!", None]

list_sum = 0
idx = 0 # added this
for value in list_mixed:
    if (type(value) != int) and (type(value) != float):
        print("Warning: Non-numerical element in position ", idx, " of list!") # change print statement
        idx = idx + 1 # added this
        continue

    list_sum = list_sum + value
    idx = idx + 1 # added this

print(list_sum)



# 1) list_mixed = [ ... ]
# =  : assignment (store right side into left)
# [] : list literal (creates a list)
# ,  : separates list items
# int: whole number, float: decimal, str: text, None: no value

# 2) list_sum = 0
# Initializes a running total (int 0).

# 3) idx = 0
# Starts a manual index counter (lists start at 0).

# 4) for value in list_mixed:
# for ... in ... : loop over items
# value          : current element in list
# :              : starts a code block (indented lines run inside loop)

# 5) if (type(value) != int) and (type(value) != float):
# if              : conditional
# type(value)     : returns exact type
# !=              : not equal
# and             : both conditions must be true
# Meaning         : treat as non-numeric if not int AND not float
# Tip            : use isinstance(value, (int, float)) for a cleaner check

# 6) print("Warning: Non-numerical element in position ", idx, " of list!")
# print(...)      : outputs to console
# multiple args   : separated by spaces
# idx             : shows position
# Tip            : f"Warning: Non-numerical element in position {idx} of list!"

# 7) idx = idx + 1
# +  : addition (increment counter)
# Shortcut: idx += 1

# 8) continue
# Skips rest of current loop and moves to next iteration.

# 9) list_sum = list_sum + value
# Adds numeric value to running total.
# Shortcut: list_sum += value

# 10) idx = idx + 1 (again)
# Increment index on numeric path too.

# 11) print(list_sum)
# Prints final total of numeric items.

# Quick reference:
# =   assignment
# []  list
# ,   separator
# :   block start
# for ... in ...  loop
# if   conditional
# ()  group condition
# and logical AND
# !=  not equal
# +   addition / increment
# continue skip iteration
# indentation defines blocks
```

<font size = "3">

In the .zip file you opened, there is a comma-separated values file named<br> "HeightWeight-quiz.csv". Using Pandas, create a DataFrame object by reading in this file.

Extract rows 345 through 397 (that is, all the rows 345, 346, 347, ..., 396, 397) of the DataFrame, saving it as another DataFrame. Then sort this subset by Height (lowest to highest), assigning it to a DataFrame called ``subset_sorted``




```python
# Write your answer here:
subset = df.iloc[345:398,:]

# acending = "True" by default
subset_sorted = subset.sort_values(by = "Height_Inches")



# 1) subset = df.iloc[345:398, :]
# df          : the original DataFrame created by reading the CSV file (e.g., pd.read_csv("HeightWeight-quiz.csv"))
# .iloc       : index-based selection (selects rows and columns by their integer positions)
# 345:398     : slice syntax (start at row 345 and stop *before* 398 → includes rows 345 through 397)
# , :         : separates row selection from column selection
# :           : selects all columns
# subset      : new DataFrame containing only rows 345–397 (inclusive)

# 2) subset_sorted = subset.sort_values(by = "Height_Inches")
# .sort_values : method to sort a DataFrame by one or more columns
# by = "Height_Inches" : specifies the column to sort by (Height column, measured in inches)
# ascending = True     : default behavior (sorts from smallest to largest)
# subset_sorted        : new DataFrame containing the subset, now sorted in ascending order by height

# Quick reference:
# df.iloc[row_start:row_end, col_start:col_end] : selects specific rows and columns by index
# :  (in row/column selection)                  : means “all”
# .sort_values(by = "ColumnName")              : sorts a DataFrame by the given column
# ascending = True/False                       : True for smallest→largest, False for largest→smallest
# =                                            : assignment (stores result into a new variable)
```

(e) Random Assignment

<font size = "3">

We want to divide the dataset into 3 groups randomly: group A, group B, group C

Randomly assign each person in the dataset one of three possible strings: ``"A"``, ``"B"``, or ``"C"``. Add a column/variable to the DataFrame which contains the letter each person is assigned. Give the column/variable the name ``"Group"``

**Hint**: Randomly choosing among 3 strings is not much different from randomly choosing among 4 numbers, which is what we did in Lecture 11

```python
# Write your answer here:
nrows, ncols = df.shape

rand_ID = np.random.choice(["A", "B", "C"], size = nrows)

df["Group"] = rand_ID


# 1) nrows, ncols = df.shape
# df.shape         : returns a tuple (number_of_rows, number_of_columns) in the DataFrame
# nrows, ncols     : unpacking the tuple into two variables
# nrows            : total number of rows (people) in the dataset
# ncols            : total number of columns (variables) in the dataset

# 2) rand_ID = np.random.choice(["A", "B", "C"], size = nrows)
# np.random.choice : randomly selects elements from a list or array
# ["A", "B", "C"]  : list of possible groups to assign
# size = nrows     : number of random selections (one for each row/person)
# rand_ID          : array of randomly chosen group labels (e.g., ["B", "A", "C", "B", ...])

# 3) df["Group"] = rand_ID
# df["Group"]      : creates a new column named "Group" in the DataFrame
# =                : assignment (stores the randomly chosen group label for each person)
# Effect           : each person is now assigned randomly to group "A", "B", or "C"

# Quick reference:
# df.shape                    : returns (rows, columns)
# np.random.choice(list, size): randomly picks 'size' number of elements from 'list'
# df["NewColumn"] = values    : adds a new column to the DataFrame
# =                           : assignment (stores data into a variable or column)
```

(f) Plotting data from one the groups

<font size = "3">

Create a new DataFrame by extracting all rows corresponding to Group B. Then make a scatter plot of height vs. weight (height on horizontal axis, weight on vertical axis) for the people in this group.

Give appropriate labels to the axes and give your scatter plot a reasonable title.

```python
# Write your answer here:

group_B = df.query("Group == 'B'")

plt.scatter(x = group_B["Height_Inches"], y = group_B["Weight_Pounds"])
plt.xlabel("Height (Inches)")
plt.ylabel("Weight (Pounds)")
plt.title("Height vs. Weight for Random Subset of Data")
plt.show()


# 1) group_B = df.query("Group == 'B'")
# df.query("condition") : filters rows based on a condition written as a string
# "Group == 'B'"       : selects only rows where the 'Group' column equals 'B'
# group_B              : new DataFrame containing only people assigned to Group B

# 2) plt.scatter(x = group_B["Height_Inches"], y = group_B["Weight_Pounds"])
# plt.scatter(x, y)    : creates a scatter plot (x → horizontal axis, y → vertical axis)
# group_B["Height_Inches"] : x-axis data (height)
# group_B["Weight_Pounds"] : y-axis data (weight)

# 3) plt.xlabel("Height (Inches)")
# plt.xlabel(...)      : sets the label for the horizontal axis

# 4) plt.ylabel("Weight (Pounds)")
# plt.ylabel(...)      : sets the label for the vertical axis

# 5) plt.title("Height vs. Weight for Random Subset of Data")
# plt.title(...)       : sets the title for the scatter plot

# 6) plt.show()
# plt.show()           : displays the figure so the scatter plot appears

# Quick reference:
# df.query("condition")             : filters rows using a condition
# ==                                : equality operator
# plt.scatter(x, y)                 : scatter plot with x on horizontal and y on vertical
# plt.xlabel("label")               : sets horizontal axis label
# plt.ylabel("label")               : sets vertical axis label
# plt.title("title")                : sets plot title
# plt.show()                        : renders the plot window
```
